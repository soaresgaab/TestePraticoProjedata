```python
import pandas as pd
import os

CAMINHO_BASE = '/content/drive/MyDrive/Mestrado UFMA/Análiticas de Aprendizagem/Dados Censo Superior/'

colunas_cursos_exigidas = ['NO_CINE_AREA_GERAL', 'NO_CURSO', 'NO_REGIAO', 'SG_UF', 'CO_IES', 'QT_ING', 'QT_ING_FEM', 'QT_MAT', 'QT_MAT_FEM', 'QT_CONC', 'QT_CONC_FEM', 'TP_MODALIDADE_ENSINO']
colunas_ies_exigidas = ['CO_IES', 'TP_CATEGORIA_ADMINISTRATIVA']

anos_para_testar = [2022, 2023, 2024] # Vamos testar apenas os que faltam

for ano in anos_para_testar:
    print(f"\n🔍 Verificando o ano de {ano}...")

    arq_cursos = f'{CAMINHO_BASE}MICRODADOS_CADASTRO_CURSOS_{ano}.CSV'

    # Lógica para o nome do ficheiro IES: tenta o antigo, se não existir, tenta o novo
    arq_ies_antigo = f'{CAMINHO_BASE}MICRODADOS_CADASTRO_IES_{ano}.CSV'
    arq_ies_novo = f'{CAMINHO_BASE}MICRODADOS_ED_SUP_IES_{ano}.CSV'

    if os.path.exists(arq_ies_antigo) or os.path.exists(arq_ies_antigo.replace('.CSV', '.csv')):
        arq_ies = arq_ies_antigo
    else:
        arq_ies = arq_ies_novo

    # Ajuste para maiúsculas/minúsculas no CSV
    if not os.path.exists(arq_cursos):
        arq_cursos = arq_cursos.replace('.CSV', '.csv')
    if not os.path.exists(arq_ies):
        arq_ies = arq_ies.replace('.CSV', '.csv')

    # Validação final de existência
    if not os.path.exists(arq_cursos):
        print(f"⚠️ Ficheiro de CURSOS de {ano} não encontrado.")
        continue
    if not os.path.exists(arq_ies):
        print(f"⚠️ Ficheiro de IES de {ano} não encontrado nem com o nome antigo nem com o novo.")
        continue

    # Lendo apenas o cabeçalho
    df_teste_cursos = pd.read_csv(arq_cursos, sep=';', encoding='latin1', nrows=0)
    df_teste_ies = pd.read_csv(arq_ies, sep=';', encoding='latin1', nrows=0)

    # Checando as colunas de Cursos
    colunas_faltantes_cursos = [col for col in colunas_cursos_exigidas if col not in df_teste_cursos.columns]
    if colunas_faltantes_cursos:
        print(f"❌ ERRO EM CURSOS {ano}: Faltam as colunas {colunas_faltantes_cursos}")
    else:
        print(f"✅ Ficheiro de Cursos {ano} OK!")

    # Checando as colunas de IES
    colunas_faltantes_ies = [col for col in colunas_ies_exigidas if col not in df_teste_ies.columns]
    if colunas_faltantes_ies:
        print(f"❌ ERRO EM IES {ano}: Faltam as colunas {colunas_faltantes_ies}")
    else:
        print(f"✅ Ficheiro de IES {ano} OK!")
```
